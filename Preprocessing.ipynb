# Removing Punvhuation

```python
sms = "Wow!!! Is this really free??? Call now."

#(a) Remove the punctuation with a regular expression.
print(re.sub(r'[^\w\s]', '', sms))

#(b) With spaCy, list the tokens that are not punctuation (is_punct). #spaCy
print([t.text for t in nlp(sms) if not t.is_punct])
```

# URLs, emails, HTML tags, @handles and #hashtags

```python
post = "Check https://uni.lk/nlp and mail admin@uni.lk #NLP @anna <b>today</b>"

# Remove the URLs
post = re.sub(r'https?://\S+|www\.\S+', '', post)
print(post)

# Remove emails
post = re.sub(r'\S+@\S+', '', post)
print(post)

# Rem0ve HTML tags
post = re.sub(r'<.*?>', '', post)
print(post)

# Remove @handles
post = re.sub(r'@\w+', '', post)
print(post)

# #NLP -> NLP
post = re.sub(r'#(\w+)', r'\1', post)

# -> Fish & chips
html.unescape("Fish &amp; chips")
```
